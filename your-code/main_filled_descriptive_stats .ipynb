# Challenge 1

#### In this challenge we will use the `Temp_States.csv` file. 

#### First import it into a data frame called `temp`.

```python
# carregar o ficheiro de temperaturas
temp = pd.read_csv('Temp_States.csv')

# olhar rapidamente as primeiras linhas para confirmar se carregou bem
temp.head()
```

#### Print `temp`

```python
# ver o dataframe completo (num projeto real eu limitaria as linhas)
temp
```

#### Select the rows where state is New York

```python
# selecionar apenas as linhas em que o estado é New York
temp_ny = temp[temp['State'] == 'New York']
temp_ny
```

#### What is the average of the temperature of cities in New York?

```python
# média da temperatura das cidades em New York
ny_avg_temp = temp_ny['Temperature'].mean()
ny_avg_temp
```

#### We want to know cities and states with Temperature above 15 degress Celcius

```python
# cidades e estados com temperatura acima de 15 ºC
hot_cities = temp[temp['Temperature'] > 15][['City', 'State', 'Temperature']]
hot_cities
```

#### Now, return only the cities that have a temperature above 15 degress Celcius

```python
# agora, só o nome das cidades com temperatura acima de 15 ºC
cities_above_15 = temp.loc[temp['Temperature'] > 15, 'City'].unique()
cities_above_15
```

#### We want to know which cities have a temperature above 15 degrees Celcius and below 20 degrees Celcius

*Hint: First write the condition then select the rows.*

```python
# condição para temperatura entre 15 e 20 ºC (exclusive)
mask_between_15_20 = (temp['Temperature'] > 15) & (temp['Temperature'] < 20)

temp_between_15_20 = temp[mask_between_15_20]
temp_between_15_20
```

```python
temp['State'].value_counts()
```

#### Find the mean and the standard deviation of the temperature of each state.

*Hint: Use functions from Data Manipulation lesson*

```python
# média e desvio padrão da temperatura por estado
temp_by_state_stats = temp.groupby('State')['Temperature'].agg(['mean', 'std'])
temp_by_state_stats
```

#### Show a visual summary of the data using boxplot. What Are the First and Third Quartiles? Comment your results.

* Hint : Quantiles vs Quartiles*
In Probability and Statistics, quantiles are cut points dividing the range of a probability distribution into continuous intervals with equal probabilities. When division is into four parts the values of the variate corresponding to 25%, 50% and 75% of the total distribution are called quartiles.

```python
# boxplot da coluna Salary para ver distribuição e outliers
employee['Salary'].plot(kind='box')
plt.ylabel('Salary')
plt.title('Boxplot dos salários')
```

```python
# primeiro quartil (25%)
q1_salary = employee['Salary'].quantile(0.25)
q1_salary
```

```python
# terceiro quartil (75%)
q3_salary = employee['Salary'].quantile(0.75)
q3_salary
```

#### Is the mean salary per gender different?

```python
# média salarial por género
mean_salary_gender = employee.groupby('Gender')['Salary'].mean()
mean_salary_gender
```

#### Find the minimum, mean and the maximum of all numeric columns for each Department.

##### Hint: Use functions from Data Manipulation lesson

```python
# mínimo, média e máximo das colunas numéricas por departamento
dept_stats = employee.groupby('Department').agg(['min', 'mean', 'max'])
dept_stats
```

### Bonus Question

#### For each department, compute the difference between the maximal salary and the minimal salary.

##### * Hint: try using `agg` or `apply` and `lambda`*

```python
# diferença entre salário máximo e mínimo por departamento
salary_range_dept = employee.groupby('Department')['Salary'].agg(lambda x: x.max() - x.min())
salary_range_dept
```

#### Open the Orders.csv dataset. Name your dataset orders

#### Explore your dataset by looking at the data types and the summary statistics. Comment your results

```python
# tipos de dados das colunas
orders.dtypes
```

```python
# estatísticas descritivas das colunas numéricas
orders.describe()
```

####  What is the average Purchase Price?

```python
# preço médio de compra (amount_spent)
avg_purchase_price = orders['amount_spent'].mean()
avg_purchase_price
```

#### What were the highest and lowest purchase prices? 

```python
# maior e menor valor de amount_spent
max_purchase = orders['amount_spent'].max()
min_purchase = orders['amount_spent'].min()
max_purchase, min_purchase
```

```python
# linhas associadas ao maior valor de compra
orders[orders['amount_spent'] == orders['amount_spent'].max()]
```

#### Select all the customers we have in Spain

```python
# todos os clientes que temos em Espanha (CustomerID)
customers_spain = orders[orders['Country'] == 'Spain']['CustomerID'].unique()
customers_spain
```

#### How many customers do we have in Spain?
##### Hint : Use value_counts()

```python
# quantos clientes distintos temos em Espanha
n_customers_spain = orders[orders['Country'] == 'Spain']['CustomerID'].nunique()
n_customers_spain
```

#### Select all the customers who have bought more than 50 items ?

```python
# encomendas onde a quantidade comprada foi maior que 50 itens
orders_qty_above_50 = orders[orders['Quantity'] > 50]
orders_qty_above_50
```

#### Select orders from Spain that are above 50 items

```python
# encomendas de Espanha com mais de 50 itens
orders_spain_above_50 = orders[(orders['Country'] == 'Spain') & (orders['Quantity'] > 50)]
orders_spain_above_50
```

#### Select all free orders

```python
# encomendas gratuitas (preço unitário igual a zero)
free_orders = orders[orders['UnitPrice'] == 0]
free_orders
```

#### Select all orders that are 'lunch bag'
#### Hint: Use string functions

```python
# encomendas cujo produto é 'lunch bag' (case insensitive)
lunch_bag_orders = orders[orders['Description'].str.contains('lunch bag', case=False, na=False)]
lunch_bag_orders
```

#### Select all orders that are made in 2011 and are 'lunch bag'  

```python
# encomendas de 2011 que são 'lunch bag'
lunch_bag_2011 = orders[
    (orders['year'] == 2011) &
    (orders['Description'].str.contains('lunch bag', case=False, na=False))
]
lunch_bag_2011
```

#### Show the frequency distribution of the amount spent in Spain.

```python
# distribuição de frequência do amount_spent para Espanha
orders_spain = orders[orders['Country'] == 'Spain']

orders_spain['amount_spent'].hist(bins=20)
plt.xlabel('amount_spent')
plt.ylabel('Frequência')
plt.title('Distribuição do amount_spent em Espanha')
```

#### Select all orders made in the month of August

```python
# encomendas feitas em agosto (mês 8)
orders_august = orders[orders['month'] == 8]
orders_august
```

#### Select how many orders are made by countries in the month of August
##### Hint: Use value_counts()

```python
# número de encomendas por país em agosto
orders_august['Country'].value_counts()
```

#### What's the  average amount of money spent by country

```python
# média do amount_spent por país
avg_amount_by_country = orders.groupby('Country')['amount_spent'].mean()
avg_amount_by_country
```

#### What's the most expensive item?

```python
# item mais caro em termos de preço unitário
max_unit_price = orders['UnitPrice'].max()
most_expensive_items = orders[orders['UnitPrice'] == max_unit_price][['Description', 'UnitPrice']].drop_duplicates()
most_expensive_items
```

#### What was the average amount spent per year ?

```python
# média de amount_spent por ano
avg_amount_per_year = orders.groupby('year')['amount_spent'].mean()
avg_amount_per_year
```
